## Clustering quality improvement

The cells below compare the current raw-feature K-Means setup against standardized, log-transformed, outlier-filtered, and alternative-feature variants. The goal is to justify the final segmentation choice with silhouette scores and diagnostic plots.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

sns.set_theme(style='whitegrid')
plt.rcParams['figure.figsize'] = (10, 6)

base_features = ['purchase_amount_usd', 'previous_purchases', 'purchase_frequency_days']
working_df = df.copy()


def prepare_matrix(frame, feature_cols, log_cols=None):
    data = frame[feature_cols].copy()
    log_cols = log_cols or []

    for column in log_cols:
        if column in data.columns:
            data[column] = np.log1p(data[column].clip(lower=0))

    scaler = StandardScaler()
    scaled = scaler.fit_transform(data)
    return data, pd.DataFrame(scaled, columns=feature_cols, index=frame.index)


def evaluate_kmeans(frame, feature_cols, k=4, log_cols=None, random_state=42):
    cleaned = frame.dropna(subset=feature_cols).copy()
    raw_data, scaled = prepare_matrix(cleaned, feature_cols, log_cols=log_cols)
    model = KMeans(n_clusters=k, random_state=random_state, n_init=10)
    labels = model.fit_predict(scaled)
    score = silhouette_score(scaled, labels)
    return cleaned, raw_data, scaled, labels, model, score


def sweep_k(frame, feature_cols, log_cols=None, k_values=range(2, 9), model_type='kmeans'):
    cleaned = frame.dropna(subset=feature_cols).copy()
    _, scaled = prepare_matrix(cleaned, feature_cols, log_cols=log_cols)

    results = []
    for k in k_values:
        if model_type == 'kmeans':
            model = KMeans(n_clusters=k, random_state=42, n_init=10)
            labels = model.fit_predict(scaled)
            inertia = model.inertia_
        else:
            model = GaussianMixture(n_components=k, random_state=42)
            labels = model.fit_predict(scaled)
            inertia = np.nan

        results.append({
            'k': k,
            'silhouette_score': silhouette_score(scaled, labels),
            'inertia': inertia,
        })

    return pd.DataFrame(results), scaled


def remove_iqr_outliers(frame, column):
    series = frame[column].dropna()
    q1 = series.quantile(0.25)
    q3 = series.quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    filtered = frame[(frame[column] >= lower) & (frame[column] <= upper)].copy()
    return filtered, lower, upper


def plot_pca_clusters(scaled_matrix, labels, title):
    pca = PCA(n_components=2, random_state=42)
    pca_values = pca.fit_transform(scaled_matrix)
    plot_df = pd.DataFrame({
        'PC1': pca_values[:, 0],
        'PC2': pca_values[:, 1],
        'cluster': labels.astype(str),
    })

    plt.figure(figsize=(10, 7))
    sns.scatterplot(data=plot_df, x='PC1', y='PC2', hue='cluster', palette='tab10', s=55, alpha=0.8)
    plt.title(title)
    plt.xlabel(f'PC1 ({pca.explained_variance_ratio_[0]:.1%} variance)')
    plt.ylabel(f'PC2 ({pca.explained_variance_ratio_[1]:.1%} variance)')
    plt.legend(title='Cluster', bbox_to_anchor=(1.02, 1), loc='upper left')
    plt.tight_layout()
    plt.show()

    return plot_df, pca
```

```python
skew_features = working_df[base_features].skew(numeric_only=True).sort_values(ascending=False)
print('Feature skewness (positive values above 1 usually benefit from log1p):')
print(skew_features.round(3))

log_features = [feature for feature, skew_value in skew_features.items() if skew_value > 1]
print('\nLog-transforming these features before scaling:')
print(log_features if log_features else 'None')

skew_plot_df = working_df[base_features].melt(var_name='feature', value_name='value')
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
for axis, feature in zip(axes, base_features):
    sns.histplot(working_df[feature].dropna(), kde=True, ax=axis, color='#3366cc')
    axis.set_title(f'{feature} distribution')
plt.tight_layout()
plt.show()
```

```python
raw_clean = working_df.dropna(subset=base_features).copy()
raw_labels = KMeans(n_clusters=4, random_state=42, n_init=10).fit_predict(raw_clean[base_features])
raw_score = silhouette_score(raw_clean[base_features], raw_labels)

print('Raw K-Means silhouette score:', round(raw_score, 3))
print('This is the current baseline before scaling or log transforms.')
```

```python
cleaned_base, raw_base, scaled_base, labels_base, kmeans_base, score_base = evaluate_kmeans(
    working_df,
    base_features,
    k=4,
    log_cols=log_features,
)

print('Standardized K-Means silhouette score:', round(score_base, 3))
print('This improves the raw-feature baseline because StandardScaler prevents purchase_amount_usd from dominating distance calculations.')

working_df['cluster_scaled_log_k4'] = labels_base
plot_pca_clusters(scaled_base, labels_base, 'PCA view of standardized K-Means clusters')
```

```python
sweep_results, sweep_scaled = sweep_k(
    working_df,
    base_features,
    log_cols=log_features,
    k_values=range(2, 9),
    model_type='kmeans',
)

fig, ax1 = plt.subplots(figsize=(10, 6))
ax1.plot(sweep_results['k'], sweep_results['inertia'], marker='o', color='#d95f02')
ax1.set_xlabel('k')
ax1.set_ylabel('Inertia', color='#d95f02')
ax1.tick_params(axis='y', labelcolor='#d95f02')

ax2 = ax1.twinx()
ax2.plot(sweep_results['k'], sweep_results['silhouette_score'], marker='o', color='#1b9e77')
ax2.set_ylabel('Silhouette score', color='#1b9e77')
ax2.tick_params(axis='y', labelcolor='#1b9e77')

plt.title('Elbow and silhouette sweep on standardized features')
plt.show()

print(sweep_results.round(3))
print('\nBest k by silhouette:', int(sweep_results.loc[sweep_results['silhouette_score'].idxmax(), 'k']))
```

```python
feature_sets = {
    'base_3_features': ['purchase_amount_usd', 'previous_purchases', 'purchase_frequency_days'],
    'base_plus_review_rating': ['purchase_amount_usd', 'previous_purchases', 'purchase_frequency_days', 'review_rating'],
    'base_plus_discount_usage': ['purchase_amount_usd', 'previous_purchases', 'purchase_frequency_days', 'discount_applied'],
    'base_plus_review_and_discount': ['purchase_amount_usd', 'previous_purchases', 'purchase_frequency_days', 'review_rating', 'discount_applied'],
}

feature_compare_rows = []
for name, feature_cols in feature_sets.items():
    compare_frame = working_df.copy()
    if 'discount_applied' in feature_cols:
        compare_frame['discount_usage_rate'] = compare_frame['discount_applied'].map({'Yes': 1, 'No': 0}).fillna(0)
        feature_cols = [col if col != 'discount_applied' else 'discount_usage_rate' for col in feature_cols]

    if 'review_rating' in feature_cols:
        feature_cols = [col for col in feature_cols if col != 'review_rating'] + ['review_rating']

    available_cols = [col for col in feature_cols if col in compare_frame.columns]
    cleaned, _, scaled, labels, _, score = evaluate_kmeans(compare_frame, available_cols, k=4, log_cols=[col for col in log_features if col in available_cols])
    feature_compare_rows.append({
        'feature_set': name,
        'features': ', '.join(available_cols),
        'silhouette_score': score,
        'rows_used': len(cleaned),
    })

feature_compare = pd.DataFrame(feature_compare_rows).sort_values('silhouette_score', ascending=False)
print(feature_compare.round(3))
print('\nBest feature combination:', feature_compare.iloc[0]['feature_set'])
```

```python
outlier_cleaned, lower_bound, upper_bound = remove_iqr_outliers(working_df, 'purchase_amount_usd')
cleaned_outlier, _, scaled_outlier, labels_outlier, _, score_outlier = evaluate_kmeans(
    outlier_cleaned,
    base_features,
    k=4,
    log_cols=log_features,
)

removed_rows = len(working_df) - len(outlier_cleaned)
print(f'Outlier rule on purchase_amount_usd: kept rows between {lower_bound:.2f} and {upper_bound:.2f}')
print(f'Removed {removed_rows} rows out of {len(working_df)}')
print('Silhouette score after outlier removal:', round(score_outlier, 3))
print('Change vs standardized baseline:', round(score_outlier - score_base, 3))

working_df['cluster_outlier_k4'] = np.nan
working_df.loc[outlier_cleaned.index, 'cluster_outlier_k4'] = labels_outlier
```

```python
best_k = int(sweep_results.loc[sweep_results['silhouette_score'].idxmax(), 'k'])

kmeans_best_cleaned, _, scaled_best, labels_kmeans_best, _, kmeans_best_score = evaluate_kmeans(
    working_df,
    base_features,
    k=best_k,
    log_cols=log_features,
)

gmm = GaussianMixture(n_components=best_k, random_state=42)
gmm_labels = gmm.fit_predict(scaled_best)
gmm_score = silhouette_score(scaled_best, gmm_labels)

print(f'Comparison at k={best_k} on the same standardized feature matrix:')
print('K-Means silhouette score:', round(kmeans_best_score, 3))
print('GMM silhouette score:    ', round(gmm_score, 3))
print('GMM helps when the clusters are elliptical or overlapping, but K-Means often wins when the structure is more spherical.')

working_df['cluster_final_kmeans'] = labels_kmeans_best
working_df['cluster_final_gmm'] = gmm_labels
```

```python
final_summary = pd.DataFrame([
    {'step': 'Original raw K-Means', 'k': 4, 'silhouette_score': raw_score, 'notes': 'Current baseline on the cleaned CSV without scaling'},
    {'step': 'Standardized + log K-Means', 'k': 4, 'silhouette_score': score_base, 'notes': 'Scaled features; log applied where skew > 1'},
    {'step': 'Outlier-removed standardized + log K-Means', 'k': 4, 'silhouette_score': score_outlier, 'notes': 'IQR filtering on purchase_amount_usd'},
    {'step': 'Best K-Means from sweep', 'k': best_k, 'silhouette_score': kmeans_best_score, 'notes': 'Best k from elbow + silhouette sweep'},
    {'step': 'Best feature combo at k=4', 'k': 4, 'silhouette_score': feature_compare.iloc[0]['silhouette_score'], 'notes': 'Best feature mix from the comparison sweep'},
    {'step': 'GMM on same scaled features', 'k': best_k, 'silhouette_score': gmm_score, 'notes': 'Alternative clustering model'},
])

overall_best_score = float(final_summary['silhouette_score'].max())
overall_best_step = final_summary.loc[final_summary['silhouette_score'].idxmax(), 'step']

print(final_summary.sort_values('silhouette_score', ascending=False).round(3))
print('\nFinal best score:', round(overall_best_score, 3))
print('Original vs final best improvement:', round(overall_best_score - raw_score, 3))
print('Best setup:', overall_best_step)
```
