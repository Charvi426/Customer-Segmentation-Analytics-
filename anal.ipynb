```python
features = df[
    [
        'purchase_frequency_days',
        'previous_purchases',
        'purchase_amount_usd'
    ]
]
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X = scaler.fit_transform(features)
```

```python
from sklearn.cluster import KMeans

inertia = []

for k in range(1,11):

    km = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    km.fit(X)

    inertia.append(km.inertia_)
```

```python
kmeans = KMeans(
    n_clusters=4,
    random_state=42,
    n_init=10
)

df['cluster'] = kmeans.fit_predict(X)
```

```python
kmeans_profile = (
    df.groupby('cluster')
    .agg({
        'purchase_amount_usd':'mean',
        'previous_purchases':'mean',
        'purchase_frequency_days':'mean'
    })
)

print(kmeans_profile)
```

```python
cluster_count = df['cluster'].value_counts().sort_index()

print("\nK-Means Cluster Distribution")
print(cluster_count)
```

```python
cluster_revenue = (
    df.groupby('cluster')['purchase_amount_usd']
      .sum()
      .sort_values(ascending=False)
)

print("\nRevenue by K-Means Cluster")
print(cluster_revenue)
```

```python

cluster_profile = (
    df.groupby('cluster')
      .agg({
          'purchase_amount_usd':'mean',
          'previous_purchases':'mean',
          'purchase_frequency_days':'mean'
      })
      .round(2)
)

print("\nK-Means Cluster Profile")
print(cluster_profile)
```

```python
cluster_revenue_share = (
    df.groupby('cluster')['purchase_amount_usd']
      .sum()
      .div(df['purchase_amount_usd'].sum())
      .mul(100)
      .round(2)
)

print("\nK-Means Revenue Share (%)")
print(cluster_revenue_share)
```

```python
cluster_subscription = pd.crosstab(
    df['cluster'],
    df['subscription_status'],
    margins=True
)

print("\nK-Means vs Subscription")
print(cluster_subscription)
```

```python
cluster_discount = pd.crosstab(
    df['cluster'],
    df['discount_applied'],
    normalize='index'
).round(3) * 100

print("\nK-Means Discount Usage (%)")
print(cluster_discount)
```

```python
cluster_rating = (
    df.groupby('cluster')['review_rating']
      .mean()
      .round(2)
)

print("\nAverage Review Rating by Cluster")
print(cluster_rating)
```

```python
from sklearn.metrics import silhouette_score

score = silhouette_score(X, df['cluster'])

print("Silhouette Score:", round(score,3))
```
